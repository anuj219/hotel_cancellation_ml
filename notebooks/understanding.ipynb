```python
# Dropping cols
data1 = data1.drop(columns=[
    "reservation_status",
    "reservation_status_date",
    "company",
    "agent"
])

# Selected Imputation
# data1["agent"] = data1["agent"].fillna(0)
data1["children"] = data1["children"].fillna(0)
data1["country"] = data1["country"].fillna(data1["country"].mode()[0])

data1.loc[data1['meal'] == 'Undefined', 'meal'] = 'Sc'
```

```python
# months 
# CYCLIC Encoding
# first map to month numbers, then convert that number of each into sin + cos coords, as sin + cos gives coords that can be used to represent all in 
# cyclic encoding (circle)
# cause standard equation of circle - 
# x = cos(angle)
# y = sin(angle)

import numpy as np

month_map = {
    "January":1,
    "February":2,
    "March":3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    'September': 9,
    "October": 10,
    "November": 11,
    "December": 12
}

data1["arrival_date_month"] = data1["arrival_date_month"].map(month_map)

data1["month_sin"] = np.sin(
    2*np.pi*data1["arrival_date_month"]/12
)
data1["month_cos"] = np.cos(
    2*np.pi*data1["arrival_date_month"]/12
)

data.drop(columns=["arrival_date_month"], inplace=True)
```

```python
# Reducing countries to 20
top20 = data1["country"].value_counts().nlargest(20).index
data1["country"] = data1["country"].apply(
    lambda x: x if x in top20 else "Other"
)
```

```python
# Shifting target to end
data1['is_canceled'] = data1.pop('is_canceled')
```

```python
# Train test split
from sklearn.model_selection import train_test_split

X = data1.iloc[::, :-1]
Y = data1.iloc[::, -1]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, random_state=32, test_size=0.3)
```

```python
# Preprocessing
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_cols = ['hotel', 'meal', 'market_segment', 'distribution_channel', 'deposit_type', 'customer_type']
numerical_cols = [
    "lead_time",
    "arrival_date_week_number",
    "arrival_date_day_of_month",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adults",
    "children",
    "babies",
    "previous_cancellations",
    "previous_bookings_not_canceled",
    "booking_changes",
    "days_in_waiting_list",
    "adr",
    "required_car_parking_spaces",
    "total_stays",
    "total_people",
    "month_sin", "month_cos"
]
# These are all columns that will be passed through the pipeline, via imputation, scaling, encoding (OHE)
# Specific columns have already been processed for this above

# data_encoded = pd.get_dummies(X_train, columns=['hotel', 'meal', 'market_segment', 'distribution_channel', 'deposit_type', 'customer_type'], dtype=int, drop_first=True)
# this is only for EDA. 
# for real ML pipeline, OneHotEncoder is better as , suppose if unseen class comes up in testing set, then OneHotEncoder wont throw an error
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),  # even if no imputers, its better to keep the imputer for the pipeline
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")), # i.e mode
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numerical_cols),
    ("cat", categorical_pipeline, categorical_cols)
])
```

--------
--------

```python
# Model Building
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=1000)
```

```python
from sklearn.pipeline import Pipeline

# No need for this
# preprocessor.fit(X_train)
# X_train_processed = preprocessor.transform(X_train)
# X_test_processed = preprocessor.transform(X_test)

pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=1000))
])
pipeline.fit(X_train, Y_train)

y_pred = pipeline.predict(X_test)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

print(classification_report(Y_test, y_pred))
print(f1_score(Y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier

pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1))
])
pipeline.fit(X_train, Y_train)

y_pred = pipeline.predict(X_test)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
print(classification_report(Y_test, y_pred))
print(f1_score(Y_test, y_pred))
```

```python
# checking feature importance for RF model
# to understand whether what feature was given what rating

feature_names = pipeline.named_steps["preprocessor"].get_feature_names_out()

importances = pipeline.named_steps["model"].feature_importances_

importance_df = (
    pd.DataFrame({
        "Feature": feature_names,
        "Importance": importances
    })
    .sort_values("Importance", ascending=False)
)

importance_df.head(20)
```

```python
# Hyperparamter tuning
param_dist = {
    "model__n_estimators": [100, 200, 300, 500],
    "model__max_depth": [10, 20, 30, None],
    "model__min_samples_split": [2, 5, 10],
    "model__min_samples_leaf": [1, 2, 4],
    "model__max_features": ["sqrt", "log2"]
}
# Ranges for parameters
# Model step was named 'model' (thus, used with double underscore notation above)

from sklearn.model_selection import RandomizedSearchCV
search = RandomizedSearchCV(
    estimator=pipeline,   # means tuning entire pipeline
    param_distributions=param_dist,
    n_iter=10, # try random 20 possible combinations, of parameters
    scoring="f1", #
    cv=5,  # 5-fold cross-validation, training data split into 5 parts, trains 5 times
    # 4 fold -> training, 1 fold -> validation, then avg f1-score
    verbose=2, # prints progress
    random_state=42, # someone else runs with same seed, they will get same 20 sample combinations
    n_jobs=1 # -1 means use all cpu cores, speeds it up | 1 means ??
)   
        
search.fit(X_train, Y_train)
print(search.best_params_) # which params performed best during cv
print(search.best_score_) # avg cv f1_score
```

```python
# Use the tuned model on the test set
best_pipeline = search.best_estimator_
y_pred = best_pipeline.predict(X_test)

print(classification_report(Y_test, y_pred))
print(f1_score(Y_test, y_pred))
```

```python
# Confusion matrix
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    Y_test,
    y_pred,
    cmap="Blues"
)

plt.title("Confusion Matrix")
plt.show()
```

```python
# ROC curve for optimal threshold finding
y_prob = best_pipeline.predict_proba(X_test)[:,1]    # 1 -> gives prob of class at COLUMN index 1, which is cancelled

from sklearn.metrics import roc_curve
from sklearn.metrics import roc_auc_score

fpr, tpr, thresholds = roc_curve(Y_test, y_prob)

auc = roc_auc_score(Y_test, y_prob)

print(auc)
```

```python
plt.figure(figsize=(7,6))

plt.plot(fpr, tpr, label=f"AUC = {auc:.3f}")

plt.plot([0,1],[0,1],"--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve")

plt.legend()

plt.show()
```

```python
# Threshold optimization
y_prob = best_pipeline.predict_proba(X_test)[:,1]    # 1 -> gives prob of class at COLUMN index 1, which is cancelled

thresholds = np.arange(0.10, 0.91, 0.01)
scores = []

for t in thresholds:
    y_pred = (y_prob >= t).astype(int)
    score = f1_score(Y_test, y_pred)
    scores.append(score)

    # Only the decision boundary changes in each loop. model remains same

best_index = np.argmax(scores)
best_threshold = thresholds[best_index]
best_f1 = scores[best_index]

print(best_threshold)
print(best_f1)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.plot(thresholds, scores)

plt.xlabel("Threshold")

plt.ylabel("F1 Score")

plt.title("Threshold vs F1 Score")

plt.grid(True)

plt.show()
```

```python
# using the optimum threshold on test data now
y_pred = (y_prob >= best_threshold).astype(int) # i.e 0.399...
print(classification_report(Y_test, y_pred))
# ConfusionMatrixDisplay.from_predictions(...)
```

```python
# Extract the fitted preprocessor from the pipeline
preprocessor = best_pipeline.named_steps["preprocessor"]

# Transform the test set for the explainer
X_test_processed = preprocessor.transform(X_test)

print(X_test_processed.shape)
```

```python
# SHAP
import shap

idx = np.random.choice(X_test_processed.shape[0], 500, replace=False)
X_sample = X_test_processed[idx]        
```

```python
rf = best_pipeline.named_steps["model"]
print("rf done")

explainer = shap.TreeExplainer(rf)
print("explainer done")

shap_values = explainer(X_sample)
print("last done")
```

```python
shap_values_class1 = shap_values[:, :, 1]

print(shap_values_class1.shape)
```

```python
# Global
shap.summary_plot(
    shap_values_class1,
    X_sample,
    feature_names=best_pipeline.named_steps["preprocessor"].get_feature_names_out()
)
    # How does this differ from feature importance -> RF Feature importance is global model-level importance measure
                                    # also it doesn't tell you why a particular booking was predicted as cancelled
    # SHAP goes further, like whether a particular lead time pushes a prediction toward cancellation or away from cancellation

i = 20

# 6. Local explanation  ... SHAP waterfall
shap.plots.waterfall(
    shap.Explanation(
        values=shap_values_class1[i],
        base_values=explainer.expected_value[1],
        data=X_sample[i],
        feature_names=feature_names
    )
)
```
